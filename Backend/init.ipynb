```python
import pandas as pd
import numpy as np
import matplotlib_inline as plt
```

```python
df = pd.read_csv("Data/emails.csv")
print(df.shape)
print(df.head())
print(df.columns)
```

```python

# 1. Dataset overview
print("Shape:", df.shape)
print("\nData types:")
print(df.dtypes)

# 2. Missing values
print("\nMissing values:")
print(df.isnull().sum())

# 3. Class distribution
print("\nSpam label distribution:")
print(df["spam"].value_counts())

print("\nClass proportions (%):")
print((df["spam"].value_counts(normalize=True) * 100).round(2))

# 4. Duplicate emails
print("\nDuplicate rows:", df.duplicated().sum())

# 5. Sample emails from each class
print("\nSpam example:")
print(df.loc[df["spam"] == 1, "text"].iloc[0][:500])

print("\nNon-spam example:")
print(df.loc[df["spam"] == 0, "text"].iloc[0][:500])
```
